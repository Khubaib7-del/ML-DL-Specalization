```python
import numpy as np 
import matplotlib.pyplot as plt
# plt.style.use("./deeplearning.mpstyle")
```

```python
# x_train is the input variables (size in 100 square feet )
# y_train is the target variables ( price in 1000s of dollars)
x_train=np.array([1.0,2.0])
y_train=np.array([300.0,500.0])
print(f"x_train = {x_train}")
print(f"y_train = {y_train}")
```

```python
# m is the number of training examples or 
# number of sample points
print(f"x_train.shape: {x_train.shape}")
m=x_train.shape[0]
print(f"Number of training examples is: {m}")
```

```python
# m is the number of total training examples
m=len(x_train)
print(f"Number of training examples is: {m}")
```

```python
i=0 # Change this to 1 to see (x^1,y^1)

x_i=x_train[i]
y_i=y_train[i]
print(f"(x^({i}),(y^({i})) = ({x_i}, {y_i})")
```

```python
# Plot the data points
plt.scatter(x_train,y_train,marker='o',c='g')
# Set the title
plt.title("Housing Prices")
# Set the y-axis
plt.ylabel("Price (in 1000s of dollars)")
# Set the x-axis
plt.xlabel("Size (1000 sqft)")
plt.show()
```

```python
# w = weight choosen by the developer
w=200
# b = bais choosen by the developer
b=100
print(f"w: {w}")
print(f"b: {b}")
```

```python
def compute_model_output(x,w,b):
    """
    Compute the prediction of a linear model
    Args:
    x (ndarray (m,)): Data, m examples
    w,b (scalar)    : model parameters
    Returns
       f_wb (ndarray (m,)): model prediction 
    """
    m= x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i]= w * x[i] + b

    return f_wb    
```

```python
tmp_f_wb=compute_model_output(x_train,w,b)

# Plot Our Model Prediction
plt.plot(x_train,tmp_f_wb,c="b",label="Our predictions")

# Plot the data points
plt.scatter(x_train,y_train,marker='x',c='r',label='Actual Values')

# Set the title
plt.title("Housing Prices")
# Set the y-axis label
plt.ylabel("Price (in 1000s of dollars)")
# Set the x-axis label
plt.xlabel("Size (in 1000 sqft)")
plt.legend()
plt.show()
```
